```python
from google.colab import files
import zipfile
from pathlib import Path
import pandas as pd

uploaded = files.upload()  # Upload CaloDiffusion_Followup_Results.zip
zip_path = next(name for name in uploaded if name.endswith(".zip"))

out = Path("/content/calo_followup")
with zipfile.ZipFile(zip_path) as z:
    z.extractall(out)

data = out / "calodiffusion_followup_results"

cells = pd.read_csv(data / "extreme_shower_all_native_cells.csv")

# Show all active cells in the unusual shower
active = cells[cells["energy_kept_GeV"] > 0].sort_values(
    "energy_kept_GeV", ascending=False
)

display(active[[
    "cell_index", "layer_id", "radial_bin",
    "r_low_XML", "r_high_XML", "energy_kept_GeV"
]].head(20))

# Summarise energy in each detector layer
display(cells.groupby("layer_id").agg(
    energy_GeV=("energy_kept_GeV", "sum"),
    active_cells=("energy_kept_GeV", lambda x: (x > 0).sum())
))
```
